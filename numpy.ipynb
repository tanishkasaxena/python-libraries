```python
#ARITHMETIC FUNCTIONS
var = np.array([1,6,8,3,78,3,0])

#tells min variable and it's position in array
print("min : ",np.min(var),np.argmin(var)) 
#tells max variable and it's position in array
print("max : ", np.max(var),np.argmax(var))
#finding square root
print("sqrt: ",np.sqrt(var))
```
